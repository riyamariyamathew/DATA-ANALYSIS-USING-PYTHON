```python
df['posted_at'] = pd.to_datetime(
    df['posted_at'],
    format='%d-%m-%Y',
    errors='coerce'
)
```

```python
df['engagement_score'] = (
    df['likes'] + df['comments'] + df['shares']
)

daily_engagement = (
    df.groupby('posted_at')['engagement_score']
    .mean()
    .reset_index()
)
```

```python
fig = px.line(
    daily_engagement,
    x='posted_at',
    y='engagement_score',
    title='Daily Average Engagement Trend',
    labels={
        'posted_at': 'Date',
        'engagement_score': 'Average Engagement'
    },
    markers=True
)

fig.show()
```

```python
#BRAZIL HAS THE HIGHEST AVERAGE ENGAGEMENT RATE
```

```python
bins = [12, 17, 24, 34, 44, 54, 64]

labels = [
    '13-17', '18-24', '25-34',
    '35-44', '45-54', '55-64'
]

df['age_group'] = pd.cut(
    df['age'],
    bins=bins,
    labels=labels
)

age_analysis = df.groupby('age_group', observed=False).agg(
    Average_Engagement=('engagement_score', 'mean'),
    Average_Engagement_Rate=('engagement_rate', 'mean')
)

print(age_analysis.round(2))
```

```python
verified_analysis = df.groupby('is_verified').agg(
    Total_Records=('user_id', 'count'),
    Average_Engagement=('engagement_score', 'mean'),
    Average_Engagement_Rate=('engagement_rate', 'mean'),
    Average_Likes=('likes', 'mean'),
    Average_Comments=('comments', 'mean'),
    Average_Shares=('shares', 'mean')
)

print(verified_analysis.round(2))
```

```python
#THERE IS NO SUCH PARTICULAR TIME OF THE DAY TO MAKE MORE IMPRESSION. NO SUCH COLUMN IS AVAILABLE IN THE DATA SET
```

```python
device_analysis = df.groupby('device_type').agg(
    Total_Records=('watch_time_sec', 'count'),
    Average_Watch_Time=('watch_time_sec', 'mean'),
    Median_Watch_Time=('watch_time_sec', 'median')
)

print(device_analysis.round(2))
```

```python
#MOBILE HAS THE HIGHEST AVERAGE DEVICE WATCH TIME
```

```python
sentiment_analysis = df.groupby('sentiment').agg(
    Total_Posts=('sentiment', 'count'),
    Average_Engagement=('engagement_score', 'mean'),
    Average_Engagement_Rate=('engagement_rate', 'mean'),
    Average_Likes=('likes', 'mean'),
    Average_Comments=('comments', 'mean'),
    Average_Shares=('shares', 'mean')
)

print(sentiment_analysis.round(2))
```

```python
#NEGATIVE SENTIMENTS PERFORMS THE BEST
```

```python
negative_neutral = df[
    df['sentiment'].isin(['negative', 'neutral'])
]

behavior_analysis = negative_neutral.groupby('sentiment').agg(
    Average_Likes=('likes', 'mean'),
    Average_Comments=('comments', 'mean'),
    Average_Shares=('shares', 'mean'),
    Average_Impressions=('impression_count', 'mean'),
    Average_Watch_Time=('watch_time_sec', 'mean'),
    Average_Engagement_Rate=('engagement_rate', 'mean')
)

print(behavior_analysis.round(2))
```
